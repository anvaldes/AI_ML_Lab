```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import NMF
```

```python
V = np.array([
    [5, 3, 0, 1, 0, 0],
    [4, 0, 0, 1, 1, 0],
    [1, 1, 0, 5, 4, 0],
    [0, 0, 0, 4, 4, 5],
    [0, 0, 0, 0, 5, 4],
    [0, 0, 5, 4, 0, 0],
])
```

```python
n_components = 2
```

```python
model = NMF(n_components=n_components, init='random', random_state=0)
W = model.fit_transform(V)
H = model.components_
```

```python
V_approx = W @ H
```

```python
fig, axs = plt.subplots(1, 4, figsize=(15, 4))
axs[0].imshow(V, cmap='viridis')
axs[0].set_title('Original Matrix V')
axs[0].axis('off')

axs[1].imshow(W, cmap='viridis')
axs[1].set_title('Matrix W (Components)')
axs[1].axis('off')

axs[2].imshow(H, cmap='viridis')
axs[2].set_title('Matrix H (Weights)')
axs[2].axis('off')

axs[3].imshow(V_approx, cmap='viridis')
axs[3].set_title('Reconstruction V_approx')
axs[3].axis('off')

plt.show()
```

```python
# Dimensionality Reduction
plt.figure(figsize=(6,5))
plt.scatter(W[:, 0], W[:, 1], c='blue', s=100)

for i in range(W.shape[0]):
    plt.text(W[i, 0]+0.02, W[i, 1]+0.02, f'Row {i}', fontsize=9)

plt.title('Dimensionality Reduction NMF (Matrix W)')
plt.xlabel('Component 1')
plt.ylabel('Component 2')
plt.grid(True)
plt.show()
```
