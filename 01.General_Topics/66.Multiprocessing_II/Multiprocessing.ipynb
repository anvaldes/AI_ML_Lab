```python
import time
import multiprocessing
from multiprocessing import Process
```

```python
names = [str(i) for i in range(1, 100 + 1)]
```

```python
def function(name):
  print('Name', name)
  time.sleep(3)
  print('-'*10)
```

### 1. Single CPU

```python
start = time.time()
```

```python
for name in names:
  function(name)
```

```python
end = time.time()
```

```python
delta = round((end - start), 2)
```

```python
print(delta)
```

### 2. Multiples CPU

```python
start = time.time()
```

```python
if __name__ == "__main__":  # confirms that the code is under main function
    procs = []
    proc = Process(target=function)  # instantiating without any argument
    procs.append(proc)
    proc.start()

    # instantiating process with arguments
    for name in names:
        # print(name)
        proc = Process(target=function, args=(name,))
        procs.append(proc)
        proc.start()

    # complete the processes
    for proc in procs:
        proc.join()
```

```python
end = time.time()
```

```python
delta = round((end - start), 2)
```

```python
print(delta)
```
